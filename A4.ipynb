Overview & Research Questions/Hypotheses
My research question is: Do shorter TikTok videos receive higher engagement rates than longer videos?

I predict that shorter videos will have higher engagement rates because people might be more likely to interact with shorter content. I will compare videos under 15 seconds, between 15 and 60 seconds and over 60 seconds. I will calculate the engagement rate by adding likes, comments, and shared, dividing by views and then multiplying by 100.
